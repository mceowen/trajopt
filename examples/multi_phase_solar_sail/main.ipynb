## Multi-Phase Solar Sail

Solves one algorithm file at a fixed revolution count, with no revolution search. Change `ALGORITHM` to run a different one:

| `ALGORITHM` | What it runs |
|---|---|
| `algorithms/ms_term_fo.yaml` | Multiple shooting, `autoscvx-term` |
| `algorithms/ms_to_hp_term_fo.yaml` | Multiple shooting → hp, `autoscvx-term` |
| `algorithms/p_to_ms_term_fo.yaml` | $p$ → multiple shooting, `autoscvx-term` |
| `algorithms/p_term_fo.yaml` | $p$, `autoscvx-term` |
| `algorithms/ms_fo.yaml` | Multiple shooting, `autoscvx` |
| `algorithms/p_fo.yaml` | $p$, `autoscvx` |

Results go to `results/<algorithm name>/`, replacing any earlier run of that algorithm, and `first_order_results.ipynb` compares them.

```python
from pathlib import Path

from IPython.display import Image, Markdown, display

from algorithm import Algorithm
from search import solve_once
from solution import print_validation_report, save_solution
from visualization import render_all

ALGORITHM = "algorithms/ms_term_fo.yaml"
REVOLUTIONS = 21

algorithm = Algorithm(Path(ALGORITHM).resolve())
OUTPUT = Path(algorithm.output_settings().directory)
print(algorithm.describe({"problem": {"revolutions": REVOLUTIONS}}))
```

```python
attempt, problem = solve_once(algorithm, REVOLUTIONS, {"problem": {"revolutions": REVOLUTIONS}}, quiet=True)
attempt.run.print_summary()
```

```python
report = attempt.report
if report is None:
    print("The final stage produced no trajectory, so there is nothing to report or plot.")
else:
    print_validation_report(report)
    save_solution(attempt.solution, OUTPUT / algorithm.output_settings().solution_file)
```

```python
if report is not None:
    render_all(attempt.solution, OUTPUT, report)
    for figure in ("summary", "transfer_plane", "elements", "control", "phases", "convergence"):
        display(Markdown(f"### {figure}"))
        display(Image(filename=OUTPUT / f"{figure}.png", width=900))
```
