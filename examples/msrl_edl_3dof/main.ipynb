## MSRL multiphase EDL

Three-phase Mars entry, descent, and landing of a notional MSRL-like vehicle (guided entry, parachute descent, powered descent) with linked phase boundaries, solved with AutoSCvx (dynamics imposed exactly, terminal and path constraints buffered): the nominal EDL case of the dissertation's multiphase EDL chapter. The guided-entry and entry-descent formulations of the same chapter are the `msrl_full_guided_entry_3dof` and `msrl_entry_descent_3dof` examples.

```python
from trajopt.trajectory_analyzer import TrajectoryAnalyzer

traj = TrajectoryAnalyzer("config.yaml")

traj.solve()
```

```python
data = traj.analyze()
```

```python
traj.plot(data, show=True)
```
